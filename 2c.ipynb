```python
text = [
    ['A','B','C','D','E','F','G','H','I','J'],
    ['B','C','D','E','F','G','H','I','J','K'],
    ['C','D','E','F','G','H','I','J','K','L'],
    ['D','E','F','A','B','C','D','E','F','G'],
    ['E','F','G','D','E','F','G','H','I','J'],
    ['F','G','H','G','H','I','J','K','L','M'],
    ['G','H','I','J','K','L','M','N','O','P'],
    ['H','I','J','K','L','M','N','O','P','Q'],
    ['I','J','K','L','M','N','O','P','Q','R'],
    ['J','K','L','M','N','O','P','Q','R','S']
]

pattern = [
    ['A','B','C'],
    ['D','E','F'],
    ['G','H','I']
]

comparisons = 0
matches = []

rows = len(text)
cols = len(text[0])

for i in range(rows - 2):
    for j in range(cols - 2):

        found = True

        for x in range(3):
            for y in range(3):

                comparisons += 1

                if text[i + x][j + y] != pattern[x][y]:
                    found = False
                    break

            if not found:
                break

        if found:
            matches.append((i, j))

print("Text Matrix")
for row in text:
    print(row)

print("\nPattern Matrix")
for row in pattern:
    print(row)

print("\nPattern Found At:", matches)
print("Total Comparisons:", comparisons)
```
