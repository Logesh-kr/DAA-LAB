```python
import random
import math

random.seed(42)

# ----------------------------
# Generate 10 random cities
# ----------------------------
N = 10

cities = [(random.randint(0, 100), random.randint(0, 100))
          for _ in range(N)]

# Distance matrix
dist = [[0] * N for _ in range(N)]

for i in range(N):
    for j in range(N):
        if i != j:
            dist[i][j] = round(math.sqrt(
                (cities[i][0] - cities[j][0]) ** 2 +
                (cities[i][1] - cities[j][1]) ** 2), 2)


# ----------------------------
# Tour Cost
# ----------------------------
def tour_cost(tour):
    cost = 0
    for i in range(len(tour) - 1):
        cost += dist[tour[i]][tour[i + 1]]
    return round(cost, 2)


# ----------------------------
# Nearest Neighbour
# ----------------------------
def nearest_neighbour():
    visited = [False] * N
    visited[0] = True

    tour = [0]
    current = 0

    for _ in range(N - 1):
        nxt = min(
            (j for j in range(N) if not visited[j]),
            key=lambda x: dist[current][x]
        )
        visited[nxt] = True
        tour.append(nxt)
        current = nxt

    tour.append(0)
    return tour


# ----------------------------
# 2-opt Improvement
# ----------------------------
def two_opt(tour):

    improved = True

    while improved:
        improved = False

        for i in range(1, len(tour) - 2):
            for j in range(i + 1, len(tour) - 1):

                if j - i == 1:
                    continue

                new_tour = (
                    tour[:i] +
                    tour[i:j][::-1] +
                    tour[j:]
                )

                if tour_cost(new_tour) < tour_cost(tour):
                    tour = new_tour
                    improved = True

    return tour


# ----------------------------
# Held-Karp DP (Optimal)
# ----------------------------
from functools import lru_cache

@lru_cache(None)
def tsp(mask, pos):

    if mask == (1 << N) - 1:
        return dist[pos][0]

    ans = float("inf")

    for city in range(N):
        if not (mask & (1 << city)):
            ans = min(
                ans,
                dist[pos][city] +
                tsp(mask | (1 << city), city)
            )

    return ans


# ----------------------------
# Solve
# ----------------------------
nn_tour = nearest_neighbour()
nn_cost = tour_cost(nn_tour)

opt_tour = two_opt(nn_tour.copy())
opt2_cost = tour_cost(opt_tour)

optimal_cost = round(tsp(1, 0), 2)

print("Random City Coordinates")
for i, c in enumerate(cities):
    print(f"City {i}: {c}")

print("\nNearest Neighbour Tour")
print(nn_tour)
print("Cost =", nn_cost)

print("\n2-opt Improved Tour")
print(opt_tour)
print("Cost =", opt2_cost)

print("\nOptimal Cost (Held-Karp)")
print(optimal_cost)

gap = ((opt2_cost - optimal_cost) / optimal_cost) * 100

print(f"\n2-opt is {gap:.2f}% above optimal.")
```
