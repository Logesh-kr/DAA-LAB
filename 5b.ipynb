```python
import math

def min_max_second(arr, low, high):
    # Base case: Single element
    if low == high:
        return arr[low], math.inf, arr[low], -math.inf

    # Base case: Two elements
    if high == low + 1:
        if arr[low] < arr[high]:
            return arr[low], arr[high], arr[high], arr[low]
        else:
            return arr[high], arr[low], arr[low], arr[high]

    # Divide
    mid = (low + high) // 2

    lmin, lsecondMin, lmax, lsecondMax = min_max_second(arr, low, mid)
    rmin, rsecondMin, rmax, rsecondMax = min_max_second(arr, mid + 1, high)

    # Combine Minimums
    if lmin < rmin:
        overallMin = lmin
        secondMin = min(lsecondMin, rmin)
    else:
        overallMin = rmin
        secondMin = min(rsecondMin, lmin)

    # Combine Maximums
    if lmax > rmax:
        overallMax = lmax
        secondMax = max(lsecondMax, rmax)
    else:
        overallMax = rmax
        secondMax = max(rsecondMax, lmax)

    return overallMin, secondMin, overallMax, secondMax


# Example
arr = [12, 5, 8, 20, 15, 2, 30, 18]

mn, secondMn, mx, secondMx = min_max_second(arr, 0, len(arr) - 1)

print("Array:", arr)
print("Minimum =", mn)
print("Second Minimum =", secondMn)
print("Maximum =", mx)
print("Second Maximum =", secondMx)
```
