```python
def first_fit_decreasing(tasks, capacity):
    tasks = sorted(tasks, reverse=True)

    remaining = []
    servers = []

    for task in tasks:
        placed = False

        for i in range(len(remaining)):
            if remaining[i] >= task:
                remaining[i] -= task
                servers[i].append(task)
                placed = True
                break

        if not placed:
            remaining.append(capacity - task)
            servers.append([task])

    return servers


def best_fit_decreasing(tasks, capacity):
    tasks = sorted(tasks, reverse=True)

    remaining = []
    servers = []

    for task in tasks:
        best = -1
        min_space = float("inf")

        for i in range(len(remaining)):
            if remaining[i] >= task and (remaining[i] - task) < min_space:
                min_space = remaining[i] - task
                best = i

        if best == -1:
            remaining.append(capacity - task)
            servers.append([task])
        else:
            remaining[best] -= task
            servers[best].append(task)

    return servers


def display(name, servers, capacity):
    print(f"\n{name}")
    print("-" * 30)

    for i, server in enumerate(servers, 1):
        used = sum(server)
        print(f"Server {i}: {server}  Used = {used:.1f}GB / {capacity}GB")

    print(f"\nTotal Servers Required = {len(servers)}")


# ---------------- Main ----------------

capacity = 8

tasks = [
    1.5, 2.0, 3.5, 0.5, 4.0,
    1.0, 2.5, 3.0, 1.5, 2.0,
    0.5, 4.0, 1.0, 3.5, 2.0
]

ffd = first_fit_decreasing(tasks, capacity)
bfd = best_fit_decreasing(tasks, capacity)

display("First Fit Decreasing (FFD)", ffd, capacity)
display("Best Fit Decreasing (BFD)", bfd, capacity)
```
