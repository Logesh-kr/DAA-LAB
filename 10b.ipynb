```python
import random
import time
import sys

sys.setrecursionlimit(30000)

comparisons = 0


# ---------------- Partition ----------------
def partition(arr, low, high):
    global comparisons

    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        comparisons += 1

        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]

    return i + 1


# ---------------- Deterministic Quick Sort ----------------
def deterministic_quicksort(arr, low, high):
    if low < high:
        pi = partition(arr, low, high)

        deterministic_quicksort(arr, low, pi - 1)
        deterministic_quicksort(arr, pi + 1, high)


# ---------------- Randomized Quick Sort ----------------
def randomized_quicksort(arr, low, high):
    if low < high:

        rand = random.randint(low, high)
        arr[rand], arr[high] = arr[high], arr[rand]

        pi = partition(arr, low, high)

        randomized_quicksort(arr, low, pi - 1)
        randomized_quicksort(arr, pi + 1, high)


# ---------------- Median-of-Three ----------------
def median_of_three(arr, low, high):

    mid = (low + high) // 2

    a = arr[low]
    b = arr[mid]
    c = arr[high]

    if (a <= b <= c) or (c <= b <= a):
        return mid
    elif (b <= a <= c) or (c <= a <= b):
        return low
    else:
        return high


def median_quicksort(arr, low, high):

    if low < high:

        median = median_of_three(arr, low, high)

        arr[median], arr[high] = arr[high], arr[median]

        pi = partition(arr, low, high)

        median_quicksort(arr, low, pi - 1)
        median_quicksort(arr, pi + 1, high)


# ---------------- Run Test ----------------
def run_test(sort_fn, arr):
    global comparisons

    a = arr[:]

    comparisons = 0

    start = time.perf_counter()

    sort_fn(a, 0, len(a) - 1)

    elapsed = (time.perf_counter() - start) * 1000

    return comparisons, elapsed


# ---------------- Test Cases ----------------
N = 10000

test_cases = {
    "Random": [random.randint(1, 100000) for _ in range(N)],
    "Sorted": list(range(N)),
    "Reverse": list(range(N, 0, -1))
}


# ---------------- Comparison Table ----------------
print(f'{"Input":<10}'
      f'{"DQS":>12}'
      f'{"RQS":>12}'
      f'{"Median3":>12}')

print("-" * 46)

for name, arr in test_cases.items():

    d_comp, d_time = run_test(deterministic_quicksort, arr)

    r_comp, r_time = run_test(randomized_quicksort, arr)

    m_comp, m_time = run_test(median_quicksort, arr)

    print(f'{name:<10}'
          f'{d_comp:>12}'
          f'{r_comp:>12}'
          f'{m_comp:>12}')

print("\nExecution Time (ms)")
print(f'{"Input":<10}'
      f'{"DQS":>12}'
      f'{"RQS":>12}'
      f'{"Median3":>12}')

print("-" * 46)

for name, arr in test_cases.items():

    d_comp, d_time = run_test(deterministic_quicksort, arr)

    r_comp, r_time = run_test(randomized_quicksort, arr)

    m_comp, m_time = run_test(median_quicksort, arr)

    print(f'{name:<10}'
          f'{d_time:>12.2f}'
          f'{r_time:>12.2f}'
          f'{m_time:>12.2f}')
```
