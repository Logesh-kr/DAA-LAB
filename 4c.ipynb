```python
import heapq

def dijkstra(graph, source):
    n = len(graph)

    dist = [float('inf')] * n
    hops = [float('inf')] * n
    prev = [None] * n

    dist[source] = 0
    hops[source] = 0

    # (distance, hops, vertex)
    pq = [(0, 0, source)]

    while pq:
        d, h, u = heapq.heappop(pq)

        if d > dist[u] or (d == dist[u] and h > hops[u]):
            continue

        for v, w in graph[u]:
            newDist = d + w
            newHops = h + 1

            if (newDist < dist[v]) or (newDist == dist[v] and newHops < hops[v]):
                dist[v] = newDist
                hops[v] = newHops
                prev[v] = u
                heapq.heappush(pq, (newDist, newHops, v))

    return dist, hops, prev


def getPath(prev, target):
    path = []
    while target is not None:
        path.append(target)
        target = prev[target]
    return path[::-1]


# Graph with 6 vertices
graph = {
    0: [(1, 2), (2, 4)],
    1: [(2, 2), (3, 5), (4, 8)],
    2: [(3, 3), (4, 4)],
    3: [(5, 2)],
    4: [(5, 1)],
    5: []
}

source = 0

dist, hops, prev = dijkstra(graph, source)

print("Vertex\tDistance\tHops\tPath")
print("-" * 45)

for v in range(len(graph)):
    path = getPath(prev, v)
    print(f"{v}\t{dist[v]}\t\t{hops[v]}\t{' -> '.join(map(str, path))}")
```
