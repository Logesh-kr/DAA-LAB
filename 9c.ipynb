```python
import random

# -------------------------------
# Next Fit Decreasing (NFD)
# -------------------------------
def next_fit_decreasing(items, capacity=1.0):
    items = sorted(items, reverse=True)

    bins = []
    remaining = 0

    for item in items:
        if not bins or remaining < item:
            bins.append([item])
            remaining = capacity - item
        else:
            bins[-1].append(item)
            remaining -= item

    return bins


# -------------------------------
# First Fit Decreasing (FFD)
# -------------------------------
def first_fit_decreasing(items, capacity=1.0):
    items = sorted(items, reverse=True)

    bins = []
    remaining = []

    for item in items:
        placed = False

        for i in range(len(remaining)):
            if remaining[i] >= item:
                bins[i].append(item)
                remaining[i] -= item
                placed = True
                break

        if not placed:
            bins.append([item])
            remaining.append(capacity - item)

    return bins


# -------------------------------
# Best Fit Decreasing (BFD)
# -------------------------------
def best_fit_decreasing(items, capacity=1.0):
    items = sorted(items, reverse=True)

    bins = []
    remaining = []

    for item in items:

        best = -1
        min_space = float("inf")

        for i in range(len(remaining)):
            if remaining[i] >= item and (remaining[i] - item) < min_space:
                best = i
                min_space = remaining[i] - item

        if best == -1:
            bins.append([item])
            remaining.append(capacity - item)
        else:
            bins[best].append(item)
            remaining[best] -= item

    return bins


# -------------------------------
# Simulation
# -------------------------------
TRIALS = 100
ITEMS = 50
CAPACITY = 1.0

ffd_total = 0
bfd_total = 0
nfd_total = 0

for _ in range(TRIALS):

    items = [round(random.uniform(0.1, 0.9), 2)
             for _ in range(ITEMS)]

    ffd_total += len(first_fit_decreasing(items, CAPACITY))
    bfd_total += len(best_fit_decreasing(items, CAPACITY))
    nfd_total += len(next_fit_decreasing(items, CAPACITY))

print("After", TRIALS, "Trials")
print("-" * 35)

print(f"Average bins used by FFD : {ffd_total / TRIALS:.2f}")
print(f"Average bins used by BFD : {bfd_total / TRIALS:.2f}")
print(f"Average bins used by NFD : {nfd_total / TRIALS:.2f}")
```
