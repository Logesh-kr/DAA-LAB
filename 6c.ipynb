```python
from functools import lru_cache


# ---------- Top-Down Memoized ----------
def matrix_chain_memo(dims):
    n = len(dims) - 1

    @lru_cache(None)
    def solve(i, j):
        if i == j:
            return 0

        ans = float('inf')

        for k in range(i, j):
            cost = (solve(i, k) +
                    solve(k + 1, j) +
                    dims[i - 1] * dims[k] * dims[j])

            ans = min(ans, cost)

        return ans

    return solve(1, n)


# ---------- Bottom-Up DP ----------
def matrix_chain_dp(dims):
    n = len(dims) - 1

    dp = [[0] * (n + 1) for _ in range(n + 1)]

    for l in range(2, n + 1):
        for i in range(1, n - l + 2):
            j = i + l - 1
            dp[i][j] = float('inf')

            for k in range(i, j):
                cost = (dp[i][k] +
                        dp[k + 1][j] +
                        dims[i - 1] * dims[k] * dims[j])

                dp[i][j] = min(dp[i][j], cost)

    return dp[1][n]


# Chain of 6 matrices
dims = [30, 35, 15, 5, 10, 20, 25]

memo_cost = matrix_chain_memo(dims)
dp_cost = matrix_chain_dp(dims)

print("Top-Down Memoized Cost :", memo_cost)
print("Bottom-Up DP Cost      :", dp_cost)

if memo_cost == dp_cost:
    print("Both methods produce the SAME minimum cost.")
else:
    print("Results are different.")
```
