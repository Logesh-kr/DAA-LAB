```python
N = 8

# Possible moves of a knight
move_x = [2, 1, -1, -2, -2, -1, 1, 2]
move_y = [1, 2, 2, 1, -1, -2, -2, -1]

# Initialize board
board = [[-1 for _ in range(N)] for _ in range(N)]

def is_safe(x, y):
    return 0 <= x < N and 0 <= y < N and board[x][y] == -1

# Count possible onward moves (Warnsdorff's heuristic)
def count_onward_moves(x, y):
    count = 0
    for i in range(8):
        nx = x + move_x[i]
        ny = y + move_y[i]
        if is_safe(nx, ny):
            count += 1
    return count

def solve(move_no, x, y):
    if move_no == N * N:
        return True

    next_moves = []

    # Collect all valid next moves
    for i in range(8):
        nx = x + move_x[i]
        ny = y + move_y[i]
        if is_safe(nx, ny):
            next_moves.append((count_onward_moves(nx, ny), nx, ny))

    # Sort by minimum onward moves
    next_moves.sort()

    # Try moves
    for _, nx, ny in next_moves:
        board[nx][ny] = move_no
        if solve(move_no + 1, nx, ny):
            return True
        board[nx][ny] = -1  # Backtrack

    return False

# Start from (0,0)
board[0][0] = 0

if solve(1, 0, 0):
    print("Knight's Tour Solution:\n")
    for row in board:
        for cell in row:
            print(f"{cell:2}", end=" ")
        print()
else:
    print("No solution exists.")
```
