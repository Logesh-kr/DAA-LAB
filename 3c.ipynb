```python
INF = float('inf')

def prim_matrix(graph):
    n = len(graph)

    key = [INF] * n
    parent = [-1] * n
    mstSet = [False] * n

    key[0] = 0

    for _ in range(n):
        # Find the minimum key vertex not yet included
        min_key = INF
        u = -1

        for v in range(n):
            if not mstSet[v] and key[v] < min_key:
                min_key = key[v]
                u = v

        mstSet[u] = True

        # Update adjacent vertices
        for v in range(n):
            if graph[u][v] != 0 and not mstSet[v] and graph[u][v] < key[v]:
                key[v] = graph[u][v]
                parent[v] = u

    print("Edges in MST:")
    total = 0
    for i in range(1, n):
        print(f"{parent[i]} - {i}  Weight: {graph[i][parent[i]]}")
        total += graph[i][parent[i]]

    print("Total MST Cost:", total)


# Adjacency Matrix
graph = [
    [0, 7, 0, 5, 0, 0, 0],
    [7, 0, 8, 9, 7, 0, 0],
    [0, 8, 0, 0, 5, 0, 0],
    [5, 9, 0, 0, 15, 6, 0],
    [0, 7, 5, 15, 0, 8, 9],
    [0, 0, 0, 6, 8, 0, 11],
    [0, 0, 0, 0, 9, 11, 0]
]

prim_matrix(graph)
```
