```python
import random
import time
import math
import heapq
import sys

sys.setrecursionlimit(30000)

# ---------------- Partition ----------------
def partition(arr, low, high):
    pivot = arr[high]
    i = low - 1

    for j in range(low, high):
        if arr[j] <= pivot:
            i += 1
            arr[i], arr[j] = arr[j], arr[i]

    arr[i + 1], arr[high] = arr[high], arr[i + 1]
    return i + 1


# ---------------- Standard Quick Sort ----------------
def quick_sort(arr, low, high):
    if low < high:
        pi = partition(arr, low, high)

        quick_sort(arr, low, pi - 1)
        quick_sort(arr, pi + 1, high)


# ---------------- Heap Sort ----------------
def heap_sort_subarray(arr, low, high):

    temp = arr[low:high + 1]

    heapq.heapify(temp)

    for i in range(low, high + 1):
        arr[i] = heapq.heappop(temp)


# ---------------- IntroSort ----------------
def intro_sort(arr):

    max_depth = 2 * int(math.log2(len(arr)))

    intro_util(arr, 0, len(arr) - 1, max_depth)


def intro_util(arr, low, high, depth_limit):

    if low >= high:
        return

    # Switch to Heap Sort if recursion becomes too deep
    if depth_limit == 0:
        heap_sort_subarray(arr, low, high)
        return

    pi = partition(arr, low, high)

    intro_util(arr, low, pi - 1, depth_limit - 1)
    intro_util(arr, pi + 1, high, depth_limit - 1)


# ---------------- Timing ----------------
def run_sort(name, sort_function, arr):

    data = arr[:]

    start = time.perf_counter()

    if name == "Quick Sort":
        sort_function(data, 0, len(data) - 1)
    else:
        sort_function(data)

    elapsed = (time.perf_counter() - start) * 1000

    return elapsed


# ---------------- Test Data ----------------
N = 10000

random_array = [random.randint(1, 100000) for _ in range(N)]
sorted_array = list(range(N))


# ---------------- Comparison ----------------
print(f'{"Input":<12} {"Quick Sort(ms)":>18} {"IntroSort(ms)":>18}')
print("-" * 50)

for name, arr in [
    ("Random", random_array),
    ("Sorted", sorted_array)
]:

    quick_time = run_sort("Quick Sort", quick_sort, arr)

    intro_time = run_sort("IntroSort", intro_sort, arr)

    print(f'{name:<12} {quick_time:>18.2f} {intro_time:>18.2f}')
```
